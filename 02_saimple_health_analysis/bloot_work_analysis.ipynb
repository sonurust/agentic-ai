```python
import os
from langchain_nvidia_ai_endpoints import ChatNVIDIA


client = ChatNVIDIA(
  base_url="https://integrate.api.nvidia.com/v1",
  model="nvidia/nemotron-3.5-super-vl-120b-a12b",
  api_key=os.environ["NVIDIA_API_KEY"],
  temperature=1,
  top_p=0.95,
  max_completion_tokens=16384,


  seed=42,
)

```

```python
from langchain_core.messages import SystemMessage, HumanMessage
```

```python
system_prompt = """
You are a medical laboratory report analysis expert. Extract all test names, values, units, reference ranges, and abnormal flags accurately from the uploaded report.
Explain each parameter in simple language, identify abnormal findings, and describe their possible clinical significance without making unsupported diagnoses.
Never invent missing data; distinguish report findings from medical interpretation and recommend appropriate doctor consultation.
"""

human_prompt = f"""
Analyze the following medical report.

BLOOD REPORT:
{blood_report}

OUTPUT FORMAT:

1. PATIENT INFORMATION
2. COMPLETE TEST RESULTS TABLE
   - Test Name
   - Observed Value
   - Unit
   - Reference Range
   - Status (Normal/High/Low)
3. ABNORMAL FINDINGS
4. PARAMETER-WISE EXPLANATION IN Hindi
5. POSSIBLE CLINICAL SIGNIFICANCE
6. RECOMMENDED FOLLOW-UP TESTS
7. QUESTIONS TO DISCUSS WITH DOCTOR
8. OVERALL CLINICAL OBSERVATIONS

IMPORTANT:
- Do not skip any test.
- Preserve original values and units.
- Do not assume missing test results.
- Clearly distinguish findings from possible interpretations.
"""

messages = [
    SystemMessage(content=system_prompt),
    HumanMessage(content=human_prompt)
]
```

```python

result = client.invoke(messages)

print(result.content)
```
